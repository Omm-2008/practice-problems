```python
mydict={"marks1":23,"marks2":121,"marks3":43,"marks4":13}
total=sum(mydict.values())
mean=total/len(mydict)
print("Mean of all values",mean)
```

```python
students={"Amit":85,"rahul":92,"priya":78,"sneha":23,"neha":10}
highS=max(students,key=students.get)
print("students with highest marks:",highS)
print("highest marks:",students[highS])           
```

```python
numbers=[10,20,30,40,50,20,10]
frequency={}
for num in numbers :
    if num in frequency:
        frequency[num]=frequency[num]+1
    else:
        frequency[num]=1
print("frequency of elements",frequency)
```

```python
my_dict={"amol":20,"ravi":30,"meena":40,"sneha":50}
sort_dict=dict(sorted(my_dict.items(),key=lambda item:item[1]))
print(sort_dict)                      
```

```python
my_dict={"amol":20,"ravi":30,"meena":40,"sneha":50}
sort_dict=dict(sorted(my_dict.items(),key=lambda item:item[1],reverse=True))
print(sort_dict)                      
```
